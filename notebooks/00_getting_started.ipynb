```python
import findspark
findspark.init()

from pyspark.sql import SparkSession
from pyspark.sql.functions import col

spark = (
    SparkSession.builder
    .appName("meu-notebook")
    .master("spark://spark-master:7077")   # crucial: sem isso, roda local[*] só no container do notebook
    .getOrCreate()
)
```

```python
df = spark.read\
        .text("/opt/spark-data/sample.txt")

df.printSchema()
```

```python
df.show(truncate=False)
```

```python
df.filter(col("value").contains("workers")).show(truncate=False)
```
